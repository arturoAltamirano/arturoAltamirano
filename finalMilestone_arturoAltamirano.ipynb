```python
#imports for data manipulation
import pandas as pd
import numpy as np
from tabulate import tabulate

#imports for data visualization
import matplotlib.pyplot as plt
import plotly.express as px
from shapely.geometry import Point

#geopandas and other geographic packages/libraries
import geopandas as gpd
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
from geopy.distance import geodesic
import zipfile

def state_countyMetrics(state_name, fips_code):
    '''takes a desired state and respective FIPS (Federal Information Processing Standard) code
       and returns a merged GeoPandas frame containing geospatial data 
       and the cancer rates by county for the requested state
       
       
       we will use this frame for visualizing cnacer rates by county'''

    #get my data from the zip provided at the data source 
    df_cancer_rates = pd.read_csv(os.path.join(cwd, 'data', 'county_zip_reg.csv'), encoding='latin1')

    #filter for only those whose county field (County, State) contains state_name
    filtered_df = df_cancer_rates[df_cancer_rates['County'].str.contains(state_name, case=False, na=False)]

    #we need to group these, as the cancer rates dataset contains many entries for a single county
    #I have chosen to average them, but any aggregation function may be used, or none at all
    countyTotals = filtered_df.groupby('County').agg({
        'incidenceRate': 'mean',
        'deathRate': 'mean',
        'avgDeathsPerYear': 'mean'}).reset_index()

    #drop the parish/county designation (Acadiana Parish -> Acadiana, Harris County -> Harris) to match other formats
    countyTotals['County'] = countyTotals['County'].str.split('Parish').str[0].str.strip()
    countyTotals['County'] = countyTotals['County'].str.split('County').str[0].str.strip()

    #prevent capitilization issues by forcing to capital
    countyTotals['County'] = countyTotals['County'].str.title()

    #each shapefile corresponds to a county
    with zipfile.ZipFile(os.path.join(cwd, "data/cb_2018_us_county_500k.zip"), "r") as zip_ref:
      zip_ref.extractall("county_shapefiles")

      #you can read these shapefiles using geopandas, this creates a 'GeoFrame' which is essentially a dataframe of metadata for a shapefile
      counties = gpd.read_file(os.path.join(cwd, 'county_shapefiles','cb_2018_us_county_500k.shp'), encoding='latin1')

    #get only the state we want according to FIPS code, this will return all surveyed counties 
    la_counties = counties[counties['STATEFP'] == fips_code]

    #merge the cancer rates dataframe with our new GeoFrame, merging onto the geodata with our respective cancer metrics
    #the new frame is merged left and right on the County data from cancer dataset, and the NAME field for the geodata
    merged = la_counties.merge(countyTotals, left_on='NAME', right_on='County')
    
    #we now return our geodata, and respective cancer rates by county
    return merged

merged_Louisiana = state_countyMetrics('Louisiana','22')
```

```python
def incidence_plotter(merged, name, points_gdf):
    ''''pass the geoframe and individual desired points to highlight
        plots the incidence and death rate'''

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

    merged.plot(column='incidenceRate', cmap='seismic', linewidth=0.8,
            ax=ax1, edgecolor='0.8', legend=True, aspect=1)

    ax1.set_title(f"Incidence Rate by County in {name}", fontsize=16)
    ax1.set_axis_off()

    merged.plot(column='deathRate', cmap='seismic', linewidth=0.8,
            ax=ax2, edgecolor='0.8', legend=True, aspect=1)

    ax2.set_title(f"Death Rate by County in {name}", fontsize=16)
    ax2.set_axis_off()

    try: 
        points_gdf.plot(ax=ax1, color='black', markersize=100)
        points_gdf.plot(ax=ax2, color='black', markersize=100)

    except:
        return

    plt.show()
```

```python
df_cancer_rates = pd.read_csv(os.path.join(cwd, 'data', 'county_zip_reg.csv'), encoding='latin1')

def state_countyMetrics_single(state_name):
    '''a slightly altered version of the function above 
       designed to only returned the grouped frame for requested state'''
    
    #get my data from the zip and from the pickle data 
    df_cancer_rates = pd.read_csv(os.path.join(cwd, 'data', 'county_zip_reg.csv'), encoding='latin1')

    #filter for only those whose county field (County, State) contains state name
    filtered_df = df_cancer_rates[df_cancer_rates['County'].str.contains(state_name, case=False, na=False)]

    countyTotals = filtered_df.groupby('County').agg({
        'incidenceRate': 'mean',
        'deathRate': 'mean',
        'avgDeathsPerYear': 'mean',
        'povertyPercent': 'mean',
        'studyCount' : 'max',
        'medIncome' : 'mean',
        'popEst2015' : 'max'
    }).reset_index()

    countyTotals['County'] = countyTotals['County'].str.split('((7,9))').str[0].str.strip()
    
    #prevent capitilization issues by forcing to capital
    countyTotals['County'] = countyTotals['County'].str.title()
    
    return countyTotals

louisianaMetrics_df = state_countyMetrics_single('Louisiana')
```

```python
coordinates = list(zip(points_df['Latitude'], points_df['Longitude']))

def find_distance(county, coordinates):
    '''given a county and a set of coordinates, find the closest coordinate to the county'''
    try:
        #county lookup returns the lat lon coordinate of the county
        geolocator = Nominatim(user_agent="county_lookup")
        location = geolocator.geocode(county)

        lat, lon = (location.latitude, location.longitude)

        #for every coordinate provided, find the distance to the provided point
        distances = [geodesic((lat, lon), pt).miles for pt in coordinates]
        
        #return the distance to the closest industrial site 
        return min(distances)
    
    except:
        #in the event a county cannot be translated (rare but it does happen)
        #return NaN to safely fill the slot in the geoframe
        return np.nan

louisianaMetrics_df["Min_Distance"] = np.nan

for index, row in louisianaMetrics_df.iterrows():
    #print(row['County'])
    louisianaMetrics_df.at[index, 'Min_Distance'] = find_distance(row['County'], coordinates)
```

```python
def get_county(lat, lon):
    '''takes a latititude and longitude and returns the associated county attribute 
       from geolocators reverse functionality'''
    
    #this try catch will keep our program from crashing if there is an error 
    #nominatim is pretty reliable, but certain coordinates can cause unpredictable errors
    try:
        #we are going to interact with nominatim's api to locate our counties and plot to a visual
        geolocator = Nominatim(user_agent="geoapi")

        #geolocator.reverse or 'Reverse Geocoding' will return the nearest OSM (open street map) object to our provided lat lon coordinate
        location = geolocator.reverse((lat, lon), language='en', exactly_one=True, timeout=10)
        print(location)

        #if there is an 'address' element for our returned OSM object (this is a neccesary field - and almost all will have this)
        if 'address' in location.raw:
            address = location.raw['address']
            return address.get('county') or address.get('parish')
        
        else: 
            return None

    except Exception as e:
        print(f"Error at ({lat},{lon}): {e}")
        return None
```

## Preprocessing

#### We need our shapefile data, and our pollution metrics. Right now the coordinates of the pollution measurements are in Lat/Lon format. We will use the above function, paired with the one below to first transcode from Lat/Lon to a parish, then join this onto our pollution measurements.

#### Essentially, finding out where the coordinates are will help us bin them to parishes, and will enable us to have a better understanding of the distribution this data represents in real space.

```python
def nitrogen_samples(df, fips_code, state_name):
    '''same as the stateCounty_metrics getter and plotter above, only modified to return the nitrogen
       phosphorus dataset for measuring pollution'''
    
    #get only data for our desired state
    df_state_nitrogen = df[df['State'].str.contains(state_name, case=False, na=False)].copy()

    #acquire county from the lat lon coordinates - use fancy lambda for list iteration and string indexing
    df_state_nitrogen.loc[:, 'County'] = df_state_nitrogen.apply(
    lambda row: get_county(row['Geographical Location (Latitude)'], row['Geographical Location (Longitude)']),
    axis=1)

    #this will be the same general structure as the prior milestone portion, only with this dataset 
    #group all county entries together to get totals by county
    countyTotals = df_state_nitrogen.groupby('County').agg({
    'Nitrogen (mg/L)': 'mean',
    'Phosphorus (mg/L)': 'mean'
    }).reset_index()

    #could do some kind of evaluation here - but rather explicitly search for both in case it uses a parish system
    countyTotals['County'] = countyTotals['County'].str.split('Parish').str[0].str.strip()
    countyTotals['County'] = countyTotals['County'].str.split('County').str[0].str.strip()

    #avoid capitilization issues between strings 
    countyTotals['County'] = countyTotals['County'].str.title()

    #print(countyTotals)

    with zipfile.ZipFile(os.path.join(cwd, "data/cb_2018_us_county_500k.zip"), "r") as zip_ref:
      zip_ref.extractall("county_shapefiles")

      counties = gpd.read_file(os.path.join(cwd, 'county_shapefiles','cb_2018_us_county_500k.shp'), encoding='latin1')

    state_counties = counties[counties['STATEFP'] == fips_code]

    merged = state_counties.merge(countyTotals, left_on='NAME', right_on='County', how='left')

    return merged
```

```python
def geomerge(dataframe, features, state_fips):
    """take our frame in, desired features, and the FIPS codes we want to include in the output geoframe
        we can use the FIPS code list to filter for whatever region we want to include
        since this is a right merge on Name and County in the dataframe."""

    #important to remember the default observation of this is counties 
    shapefile = os.path.join(cwd, 'county_shapefiles', 'cb_2018_us_county_500k.shp')
    us_counties = gpd.read_file(shapefile)

    #filter only for those counties who are in our FIPS code list
    us_counties = us_counties[us_counties["STATEFP"].isin(state_fips)]

    #merging right, so we drop anythign that does not match on the frames 'County' column
    geo_merged = us_counties.merge(dataframe, left_on="NAME", right_on="County", how="left")

    #dropping any messed up rows/renaming geometry and dropping useless y dim like before
    #geo_merged = geo_merged.dropna(subset=features)
    geo_merged = geo_merged.rename(columns={'geometry_x': 'geometry'})
    geo_merged = geo_merged.drop(columns=['geometry_y'])

    #assign to a variable and set the crs 
    geo_merged = gpd.GeoDataFrame(geo_merged, geometry='geometry', crs='EPSG:4326')

    return geo_merged
```

## Milestone 3: How about we train a KNN regressor and evaluate it's performance between regions.

#### Going to build a model to predict incidence rate by pollution levels. We will use the data we have inspected so far from the greater south to predict for the entire nation. 

#### Our model will have peeked at some of our final testing data, but this dataset is so sparse I feel it is neccesary. 

#### Let's build a 10 nearest neighbors regressor to predict cancer incidence rates based on reported pollution levels from the south dataset we already have. If this proves fruitful, we can expand our training scope to the entire nation.

```python
predictions = []
actuals = []

features = ["NO2_Mean", "O3_Mean", "SO2_Mean", "CO_Mean"]
target = "incidenceRate"

#split train and test 80/20
X_train, X_test, y_train, y_test = train_test_split(
    merged_final_southern_states[features],
    merged_final_southern_states[target],
    test_size=0.2,
    random_state=444
)

#10nn regression model
model = KNeighborsRegressor(n_neighbors=10)
model.fit(X_train, y_train)

#loop through the partitioned test set
for i in range(len(X_test)):
    try:
        X_row = X_test.iloc[i].to_frame().T
        pred = model.predict(X_row)[0]
        actual = y_test.iloc[i]
        
        predictions.append(pred)
        actuals.append(actual)

    except Exception as e:
        print(f"Error on index {i}: {e}")
        pass

actuals_arr = np.array(actuals)
preds_arr = np.array(predictions)

#mean absolute percentage error (MAPE)
mape = np.mean(np.abs((actuals_arr - preds_arr) / actuals_arr)) * 100

#manhattan distance (MAE) or arithmetic average of absolute errors  ors 
mae = mean_absolute_error(actuals_arr, preds_arr)

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Absolute Percentage Error (MAPE): {mape:.2f}%")
```

## Lets train on the whole nation, and predict on individual states. 

#### Again, this will cheat slightly, but is only to serve as a baseline for future evaluation. 

#### I will be gathering more samples myself after graduation. I love to spend time on the water and will be doing so for a living after graduation so can do so at will. I will be kayaking and camping in the north of Louisana, and will sample the soil and river's for these pollutants.

```python
def trainTest_model(k, X_train, X_test, y_train, y_test):
    #knn regression model
    model = KNeighborsRegressor(n_neighbors = k)
    model.fit(X_train, y_train)

    predictions = []
    actuals = []

    #loop through the partitioned test set
    for i in range(len(X_test)):
        try:
            X_row = X_test.iloc[i].to_frame().T
            pred = model.predict(X_row)[0]
            actual = y_test.iloc[i]
        
            predictions.append(pred)
            actuals.append(actual)

        except Exception as e:
            print(f"Error on index {i}: {e}")
            pass

    actuals_arr = np.array(actuals)
    preds_arr = np.array(predictions)

    #mean absolute percentage error (MAPE)
    mape = np.mean(np.abs((actuals_arr - preds_arr) / actuals_arr)) * 100

    #manhattan distance (MAE) or arithmetic average of absolute errors
    mae = mean_absolute_error(actuals_arr, preds_arr)

    r2 = r2_score(actuals, predictions)

    return model, mape, mae, r2
```

## Parameter and Model Evalaution 

#### Lets loop over our model training method to find our best K parameter. This is the same as homeworks 9 and 10, only I will be capturing some more/different metrics.

#### Once we have our best model, we can use it to predict on individual states/new data collections for a county

#### I am going to cap this at 250, as that seems to limit runtime to a reasonable amount. If I extend the value to the size of the training set, it seems to decrease test performance despite higher overall test/train accuracy.

#### This is likely a function of overfitting, and can be artifically limited by simply restricting N

```python
model_list = []
mape_list = []
mae_list = []
r2_list = []
k_values = []

best_mae = 10000
best_mape = 100.00
best_model = KNeighborsRegressor
best_k = 0

#split train and test 80/20
X_train, X_test, y_train, y_test = train_test_split(
    merged_all_states[features],
    merged_all_states[target],
    test_size=0.2,
    random_state=444)

#this is my answer from lab 10 - simply loop over a fixed range and select the best model
#according to some criteria, and the respective k value at the time of achieving this best
for k in range(1, 200):
    model, mape, mae, r2 = trainTest_model(k, X_train, X_test, y_train, y_test)

    #capture these best values and use them for our k decision
    if mae < best_mae:
        best_mae = mae
        best_model = model
        best_k = k

    if mape < best_mape:
        best_mape = mape
    
    mape_list.append(mape)
    mae_list.append(mae)
    r2_list.append(r2)
    k_values.append(k)
```

## We have our best MAE and MAPE k value, and can use this to make a decision on our ideal K for prediction. 

#### I will keep this simple and use the best_k value captured on the back of the MAE value. But you could tune this however you like.

```python
fig, (ax1,ax2) = plt.subplots(1,2)
fig.set_figheight(10)
fig.set_figwidth(40)

#probably a fancier way of doing this iteratively but manually specifying each subplot works
ax1.plot(k_values, r2_list, linewidth=5, color='red')
ax1.set_xlabel('K-Value')
ax1.set_ylabel('MAPE')
ax1.set_title('Mean Absolute Percentage Error')

#probably a fancier way of doing this iteratively but manually specifying each subplot works
ax2.plot(k_values, mae_list, linewidth=5, color='blue')
ax2.set_xlabel('K-Value')
ax2.set_ylabel('MAE')
ax2.set_title('Mean Absolute Error')

print(f"Best Average Error: {best_mae} at K: {best_k}")
```

## Trained on whole country -> Predict on single state

#### The below function will predict on a single state, we can visualize our model's capability on a local isolated region.

#### We essentially 'go both ways' analyzing the data's ability to scale up or down in scale. 

```python
def statePredictor(state, fipsCode, model):
    merged_State = state_countyMetrics(state, fipsCode)
    merged_State = merged_State.merge(pollution_means, on="County", how="left")
    us_counties = gpd.read_file(os.path.join(cwd, 'county_shapefiles', 'cb_2018_us_county_500k.shp'))
    us_counties = us_counties[us_counties["STATEFP"] == fipsCode]

    geo_merged = us_counties.merge(merged_State, left_on="NAME", right_on="County", how="left")

    geo_merged = geo_merged.dropna(subset=features)

    #make a prediction and insert it into our dataframe for later
    geo_merged["predicted_incidence"] = model.predict(geo_merged[features])

    #rename these dummy columns to clean things up
    geo_merged = geo_merged.rename(columns={'geometry_x': 'geometry'})
    geo_merged = geo_merged.drop(columns=['geometry_y'], errors='ignore')

    #geo_merged = geo_merged[[col for col in geo_merged.columns if not col.endswith('_y')]]
    geo_merged = gpd.GeoDataFrame(geo_merged, geometry='geometry', crs='EPSG:4326')

    fig, axes = plt.subplots(1, 2, figsize=(18, 8))

    geo_merged.plot(column="incidenceRate", cmap="seismic", legend=True, ax=axes[0])
    axes[0].set_title("Actual Cancer Incidence Rate")
    axes[0].axis("off")

    geo_merged.plot(column="predicted_incidence", cmap="seismic", legend=True, ax=axes[1])
    axes[1].set_title("Predicted Cancer Incidence Rate (Model)")
    axes[1].axis("off")

    plt.tight_layout()
    plt.show()
```
